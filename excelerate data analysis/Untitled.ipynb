```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading datasets

df1 = pd.read_csv('data/UserData_cleaned.csv')
df2 = pd.read_csv('data/Opportunity Wise Data cleaned.csv')
```

```python
df1.info()
```

```python
df2.info()
```

```python
# Step 1: Reduce Memory Usage  
for col in ['PreferredSponsors', 'Gender', 'Country', 'Degree', 'city', 'isFromSocialMedia']:  
    df1[col] = df1[col].astype('category')  

for col in ['Opportunity Id', 'Opportunity Name', 'Opportunity Category', 'Gender', 'City', 'State',  
            'Country', 'Current Student Status', 'Current/Intended Major', 'Status Description',  
            'Badge Id', 'Badge Name', 'Skills Earned']:  
    df2[col] = df2[col].astype('category') 

df1['isFromSocialMedia'] = df1['isFromSocialMedia'].astype(bool)  

df2['Reward Amount'] = pd.to_numeric(df2['Reward Amount'], errors='coerce')  
df2['Skill Points Earned'] = pd.to_numeric(df2['Skill Points Earned'], errors='coerce')  

df1['Sign Up Date'] = pd.to_datetime(df1['Sign Up Date'], errors='coerce')  
df2['Opportunity End Date'] = pd.to_datetime(df2['Opportunity End Date'], errors='coerce')  
df2['Apply Date'] = pd.to_datetime(df2['Apply Date'], errors='coerce')  
df2['Opportunity Start Date'] = pd.to_datetime(df2['Opportunity Start Date'], errors='coerce')  
```

```python
# Step 2: Merge Datasets 
print(df1.columns)
print(df2.columns)

```

```python
df1.rename(columns={'city': 'City'}, inplace=True)
```

```python
print(df1.columns)
print(df2.columns)
```

```python
merged_df = df1.merge(df2, on=['Gender', 'Country', 'City'], how='left')
```

```python
merged_df.info()
```

```python
# Check for duplicates based on the key columns used in the merge
print(merged_df.duplicated(subset=['Gender', 'Country', 'City']).sum())
```
